# Archivo para el Análisis Exploratorio de los datos


```python
## 1. Importar librerías y cargar los datos
import pandas as pd
import plotly.express as px

# Cargar el dataset desde la carpeta principal del proyecto
df = pd.read_csv('../vehicles_us.csv')

# Mostrar las primeras filas
df.head()
```

```python
##2 Exploración general

# Dimensiones del dataset
print("Dimensiones:", df.shape)

# Tipos de datos y valores no nulos
print("\nINFORMACIÓN GENERAL")
df.info()

# Estadísticas descriptivas
print("\nESTADÍSTICAS DESCRIPTIVAS")
display(df.describe())

# Valores ausentes
print("\nVALORES AUSENTES")
display(df.isna().sum().sort_values(ascending=False))

# Número de registros por condición del vehículo
print("\nCONDICIÓN DE LOS VEHÍCULOS")
display(df['condition'].value_counts())
```

```python
# Histograma precios

fig = px.histogram(
    df,
    x='price',
    nbins=60,
    title='Distribución de precios de vehículos',
    labels={'price': 'Precio (USD)'},
    marginal='box'
)

fig.show()
```

```python
# Distribución de precios sin valores extremos superiores a 50,000 USD

df_filtered = df[df['price'] <= 50000].copy()

fig = px.histogram(
    df_filtered,
    x='price',
    nbins=60,
    title='Distribución de precios hasta 50,000 USD',
    labels={'price': 'Precio (USD)'}
)

fig.show()
```

```python
# Gráfico de dispersión entre kilometraje y precio

fig = px.scatter(
    df_filtered,
    x='odometer',
    y='price',
    color='condition',
    opacity=0.5,
    title='Relación entre kilometraje y precio',
    labels={
        'odometer': 'Kilometraje (millas)',
        'price': 'Precio (USD)',
        'condition': 'Condición'
    }
)

fig.show()
```

```python
# Diagrama de caja por condición del vehículo

fig = px.box(
    df_filtered,
    x='condition',
    y='price',
    color='condition',
    title='Precio de los vehículos según su condición',
    labels={
        'condition': 'Condición del vehículo',
        'price': 'Precio (USD)'
    }
)

fig.show()
```

```python
# Cantidad de anuncios por tipo de vehículo

type_counts = (
    df['type']
    .value_counts()
    .reset_index()
)

type_counts.columns = ['type', 'count']

fig = px.bar(
    type_counts,
    x='type',
    y='count',
    color='type',
    title='Cantidad de anuncios por tipo de vehículo',
    labels={
        'type': 'Tipo de vehículo',
        'count': 'Cantidad de anuncios'
    }
)

fig.show()
```

```python
# Calcular precio promedio por tipo

avg_prices = (
    df.groupby('type')['price']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    avg_prices,
    x='type',
    y='price',
    color='type',
    title='Precio promedio por tipo de vehículo',
    labels={
        'type': 'Tipo de vehículo',
        'price': 'Precio promedio (USD)'
    }
)

fig.show()
```
